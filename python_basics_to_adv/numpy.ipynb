Why numpy instead of lists ?


Lists are slow, while numpy is fast.<br>
In numpy it represents a data in an efficient memory manner. <br>
Faster to read less bytes of memory.<br>
It uses Contiguous memory. <br> 
while lists uses scattered computer's memory.<br>

Effective Cache Utitilization<br>

#### Copy Array

```python
a = np.random.randint(1, 11, size=(3,5))

b = a
a[1][1] = 35
a,b
```

```python
a = np.random.randint(1,11, size=(3,5))
b = a.copy()
a[1][1] = 35
b[1][1] = 98
a,b
```
