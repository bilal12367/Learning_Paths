```python
# String methods and examples

# 1. capitalize() - Capitalizes the first character of the string
example = "hello world"
print(example.capitalize())  # Output: "Hello world"

# 2. lower() - Converts all characters to lowercase
print(example.lower())  # Output: "hello world"

# 3. upper() - Converts all characters to uppercase
print(example.upper())  # Output: "HELLO WORLD"

# 4. title() - Converts the first character of each word to uppercase
print(example.title())  # Output: "Hello World"

# 5. strip() - Removes leading and trailing whitespace
example_with_spaces = "   hello world   "
print(example_with_spaces.strip())  # Output: "hello world"

# 6. lstrip() - Removes leading whitespace
print(example_with_spaces.lstrip())  # Output: "hello world   "

# 7. rstrip() - Removes trailing whitespace
print(example_with_spaces.rstrip())  # Output: "   hello world"

# 8. replace() - Replaces occurrences of a substring with another substring
print(example.replace("world", "Python"))  # Output: "hello Python"

# 9. split() - Splits the string into a list based on a delimiter
print(example.split())  # Output: ['hello', 'world']

# 10. join() - Joins elements of a list into a string with a specified delimiter
words = ['hello', 'world']
print(" ".join(words))  # Output: "hello world"

# 11. find() - Returns the index of the first occurrence of a substring
print(example.find("world"))  # Output: 6

# 12. index() - Similar to find(), but raises an error if the substring is not found
print(example.index("world"))  # Output: 6

# 13. count() - Counts the occurrences of a substring
print(example.count("l"))  # Output: 3

# 14. startswith() - Checks if the string starts with a specified substring
print(example.startswith("hello"))  # Output: True

# 15. endswith() - Checks if the string ends with a specified substring
print(example.endswith("world"))  # Output: True

# 16. isalpha() - Checks if all characters are alphabetic
print(example.isalpha())  # Output: False (because of the space)

# 17. isdigit() - Checks if all characters are digits
numeric_string = "12345"
print(numeric_string.isdigit())  # Output: True

# 18. isalnum() - Checks if all characters are alphanumeric
alphanumeric_string = "hello123"
print(alphanumeric_string.isalnum())  # Output: True

# 19. isspace() - Checks if the string contains only whitespace
whitespace_string = "   "
print(whitespace_string.isspace())  # Output: True

# 20. zfill() - Pads the string with zeros on the left to reach a specified width
print(numeric_string.zfill(8))  # Output: "00012345"
```

```python
import math

# 1. abs() - Returns the absolute value of a number
print(abs(-5))  # Output: 5

# 2. math.ceil() - Rounds a number up to the nearest integer
print(math.ceil(4.2))  # Output: 5

# 3. math.floor() - Rounds a number down to the nearest integer
print(math.floor(4.8))  # Output: 4

# 4. math.sqrt() - Returns the square root of a number
print(math.sqrt(16))  # Output: 4.0

# 5. math.pow() - Raises a number to the power of another
print(math.pow(2, 3))  # Output: 8.0

# 6. round() - Rounds a number to the nearest integer or specified number of decimals
print(round(3.14159, 2))  # Output: 3.14

# 7. math.log() - Returns the natural logarithm (base e) of a number
print(math.log(10))  # Output: 2.302585092994046

# 8. math.log10() - Returns the base-10 logarithm of a number
print(math.log10(100))  # Output: 2.0

# 9. math.exp() - Returns e raised to the power of a number
print(math.exp(2))  # Output: 7.38905609893065

# 10. math.sin() - Returns the sine of a number (in radians)
print(math.sin(math.pi / 2))  # Output: 1.0

# 11. math.cos() - Returns the cosine of a number (in radians)
print(math.cos(0))  # Output: 1.0

# 12. math.tan() - Returns the tangent of a number (in radians)
print(math.tan(math.pi / 4))  # Output: 1.0

# 13. math.degrees() - Converts radians to degrees
print(math.degrees(math.pi))  # Output: 180.0

# 14. math.radians() - Converts degrees to radians
print(math.radians(180))  # Output: 3.141592653589793

# 15. math.factorial() - Returns the factorial of a number
print(math.factorial(5))  # Output: 120

# 16. math.gcd() - Returns the greatest common divisor of two numbers
print(math.gcd(48, 18))  # Output: 6

# 17. math.isfinite() - Checks if a number is finite
print(math.isfinite(var1))  # Output: True

# 18. math.isinf() - Checks if a number is infinite
print(math.isinf(float('inf')))  # Output: True

# 19. math.isnan() - Checks if a number is NaN (Not a Number)
print(math.isnan(float('nan')))  # Output: True

# 20. math.copysign() - Returns a float with the magnitude of the first number and the sign of the second
print(math.copysign(5, -1))  # Output: -5.0
```

```python
# List methods and examples

# 1. append() - Adds an element to the end of the list
example_list = [1, 2, 3]
example_list.append(4)
print(example_list)  # Output: [1, 2, 3, 4]

# 2. extend() - Extends the list by appending elements from another iterable
example_list.extend([5, 6])
print(example_list)  # Output: [1, 2, 3, 4, 5, 6]

# 3. insert() - Inserts an element at a specified position
example_list.insert(2, 99)
print(example_list)  # Output: [1, 2, 99, 3, 4, 5, 6]

# 4. remove() - Removes the first occurrence of a specified value
example_list.remove(99)
print(example_list)  # Output: [1, 2, 3, 4, 5, 6]

# 5. pop() - Removes and returns the element at a specified position (default is the last element)
popped_element = example_list.pop()
print(popped_element)  # Output: 6
print(example_list)  # Output: [1, 2, 3, 4, 5]

# 6. clear() - Removes all elements from the list
example_list.clear()
print(example_list)  # Output: []

# 7. index() - Returns the index of the first occurrence of a specified value
example_list = [10, 20, 30, 20]
print(example_list.index(20))  # Output: 1

# 8. count() - Returns the number of occurrences of a specified value
print(example_list.count(20))  # Output: 2

# 9. sort() - Sorts the list in ascending order (or descending if `reverse=True`)
example_list.sort()
print(example_list)  # Output: [10, 20, 20, 30]
example_list.sort(reverse=True)
print(example_list)  # Output: [30, 20, 20, 10]

# 10. reverse() - Reverses the order of the list
example_list.reverse()
print(example_list)  # Output: [10, 20, 20, 30]

# 11. copy() - Returns a shallow copy of the list
copied_list = example_list.copy()
print(copied_list)  # Output: [10, 20, 20, 30]

# 12. len() - Returns the number of elements in the list
print(len(example_list))  # Output: 4

# 13. max() - Returns the largest element in the list
print(max(example_list))  # Output: 30

# 14. min() - Returns the smallest element in the list
print(min(example_list))  # Output: 10

# 15. sum() - Returns the sum of all elements in the list
print(sum(example_list))  # Output: 80

# 16. sorted() - Returns a new sorted list (does not modify the original list)
new_sorted_list = sorted(example_list)
print(new_sorted_list)  # Output: [10, 20, 20, 30]
print(example_list)  # Output: [10, 20, 20, 30]

# 17. enumerate() - Returns an enumerate object with index and value pairs
for index, value in enumerate(example_list):
    print(f"Index: {index}, Value: {value}")
# Output:
# Index: 0, Value: 10
# Index: 1, Value: 20
# Index: 2, Value: 20
# Index: 3, Value: 30

# 18. filter() - Filters elements based on a condition
filtered_list = list(filter(lambda x: x > 15, example_list))
print(filtered_list)  # Output: [20, 20, 30]

# 19. map() - Applies a function to all elements in the list
mapped_list = list(map(lambda x: x * 2, example_list))
print(mapped_list)  # Output: [20, 40, 40, 60]

# 20. list comprehension - A concise way to create lists
squared_list = [x ** 2 for x in example_list]
print(squared_list)  # Output: [100, 400, 400, 900]
```

```python
import math

# 1. Basic list comprehension
squared_numbers = [x ** 2 for x in range(10)]
print(squared_numbers)  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

# 2. List comprehension with a condition
even_numbers = [x for x in range(10) if x % 2 == 0]
print(even_numbers)  # Output: [0, 2, 4, 6, 8]

# 3. Nested list comprehension
matrix = [[j for j in range(3)] for i in range(3)]
print(matrix)  # Output: [[0, 1, 2], [0, 1, 2], [0, 1, 2]]

# 4. Flattening a nested list
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8]]
flattened = [item for sublist in nested_list for item in sublist]
print(flattened)  # Output: [1, 2, 3, 4, 5, 6, 7, 8]

# 5. Applying a function to elements
square_roots = [math.sqrt(x) for x in range(1, 10)]
print(square_roots)  # Output: [1.0, 1.4142135623730951, ..., 3.0]

# 6. Conditional transformation
numbers = [-5, -3, 0, 2, 4]
positive_numbers = [x if x > 0 else 0 for x in numbers]
print(positive_numbers)  # Output: [0, 0, 0, 2, 4]

# 7. Cartesian product
colors = ['red', 'green']
objects = ['ball', 'box']
cartesian_product = [(color, obj) for color in colors for obj in objects]
print(cartesian_product)  # Output: [('red', 'ball'), ('red', 'box'), ...]

# 8. Filtering with multiple conditions
filtered_numbers = [x for x in range(20) if x % 2 == 0 and x % 3 == 0]
print(filtered_numbers)  # Output: [0, 6, 12, 18]

# 9. List comprehension with enumerate
words = ['hello', 'world']
indexed_words = [(i, word) for i, word in enumerate(words)]
print(indexed_words)  # Output: [(0, 'hello'), (1, 'world')]

# 10. Using a set to remove duplicates
numbers_with_duplicates = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = [x for x in set(numbers_with_duplicates)]
print(unique_numbers)  # Output: [1, 2, 3, 4, 5]
```

```python
# Dictionary basics and functions

# 1. Creating a dictionary
example_dict = {'name': 'Alice', 'age': 25, 'city': 'New York'}
print(example_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}

# 2. Accessing values using keys
print(example_dict['name'])  # Output: Alice

# 3. Using get() to access values (returns None if key is not found)
print(example_dict.get('age'))  # Output: 25
print(example_dict.get('country'))  # Output: None

# 4. Adding a new key-value pair
example_dict['country'] = 'USA'
print(example_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York', 'country': 'USA'}

# 5. Updating a value
example_dict['age'] = 26
print(example_dict)  # Output: {'name': 'Alice', 'age': 26, 'city': 'New York', 'country': 'USA'}

# 6. Removing a key-value pair using pop()
removed_value = example_dict.pop('city')
print(removed_value)  # Output: New York
print(example_dict)  # Output: {'name': 'Alice', 'age': 26, 'country': 'USA'}

# 7. Removing the last inserted key-value pair using popitem()
last_item = example_dict.popitem()
print(last_item)  # Output: ('country', 'USA')
print(example_dict)  # Output: {'name': 'Alice', 'age': 26}

# 8. Checking if a key exists in the dictionary
print('name' in example_dict)  # Output: True
print('city' in example_dict)  # Output: False

# 9. Iterating over keys
for key in example_dict.keys():
    print(key)
# Output:
# name
# age

# 10. Iterating over values
for value in example_dict.values():
    print(value)
# Output:
# Alice
# 26

# 11. Iterating over key-value pairs
for key, value in example_dict.items():
    print(f"{key}: {value}")
# Output:
# name: Alice
# age: 26

# 12. Merging dictionaries using update()
another_dict = {'city': 'Los Angeles', 'hobby': 'reading'}
example_dict.update(another_dict)
print(example_dict)
# Output: {'name': 'Alice', 'age': 26, 'city': 'Los Angeles', 'hobby': 'reading'}

# 13. Clearing all key-value pairs
example_dict.clear()
print(example_dict)  # Output: {}

# 14. Creating a dictionary with default values using fromkeys()
keys = ['a', 'b', 'c']
default_dict = dict.fromkeys(keys, 0)
print(default_dict)  # Output: {'a': 0, 'b': 0, 'c': 0}

# 15. Copying a dictionary
original_dict = {'x': 1, 'y': 2}
copied_dict = original_dict.copy()
print(copied_dict)  # Output: {'x': 1, 'y': 2}

# 16. Using setdefault() to get a value or set it if the key does not exist
value = original_dict.setdefault('z', 3)
print(value)  # Output: 3
print(original_dict)  # Output: {'x': 1, 'y': 2, 'z': 3}

# 17. Dictionary comprehension
squared_numbers = {x: x ** 2 for x in range(5)}
print(squared_numbers)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# 18. len() - Returns the number of key-value pairs
print(len(original_dict))  # Output: 3

# 19. keys() - Returns a view object of all keys
print(original_dict.keys())  # Output: dict_keys(['x', 'y', 'z'])

# 20. values() - Returns a view object of all values
print(original_dict.values())  # Output: dict_values([1, 2, 3])

# 21. items() - Returns a view object of all key-value pairs
print(original_dict.items())  # Output: dict_items([('x', 1), ('y', 2), ('z', 3)])
```

```python
from itertools import chain

# Iterator functions and examples

# 1. iter() - Returns an iterator object
example_list = [1, 2, 3, 4]
iterator = iter(example_list)
print(next(iterator))  # Output: 1
print(next(iterator))  # Output: 2

# 2. next() - Retrieves the next item from an iterator
print(next(iterator))  # Output: 3
print(next(iterator))  # Output: 4
# If you call next() again, it will raise StopIteration

# 3. enumerate() - Returns an iterator that yields pairs of index and value
for index, value in enumerate(example_list):
    print(f"Index: {index}, Value: {value}")
# Output:
# Index: 0, Value: 1
# Index: 1, Value: 2
# Index: 2, Value: 3
# Index: 3, Value: 4

# 4. zip() - Combines multiple iterables into an iterator of tuples
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
zipped = zip(list1, list2)
print(list(zipped))  # Output: [(1, 'a'), (2, 'b'), (3, 'c')]

# 5. map() - Applies a function to all items in an iterable
squared = map(lambda x: x ** 2, example_list)
print(list(squared))  # Output: [1, 4, 9, 16]

# 6. filter() - Filters elements based on a condition
filtered = filter(lambda x: x > 2, example_list)
print(list(filtered))  # Output: [3, 4]

# 7. reversed() - Returns an iterator that accesses the given sequence in reverse order
reversed_list = reversed(example_list)
print(list(reversed_list))  # Output: [4, 3, 2, 1]

# 8. sorted() - Returns a sorted iterator
unsorted_list = [3, 1, 4, 2]
sorted_iterator = sorted(unsorted_list)
print(sorted_iterator)  # Output: [1, 2, 3, 4]

# 9. range() - Returns an iterator that generates numbers within a specified range
for i in range(5):
    print(i, end=" ")  # Output: 0 1 2 3 4

# 10. itertools.chain() - Combines multiple iterables into a single iterator
list3 = [5, 6]
chained = chain(example_list, list3)
print(list(chained))  # Output: [1, 2, 3, 4, 5, 6]
```

```python
class MyIterator:
    def __init__(self, limit):
        self.limit = limit
        self.cnt = 0
    def __iter__(self):
        return self

    def __next__(self):
        if(self.cnt < self.limit):
            self.cnt += 1
            return self.cnt
        else:
            raise StopIteration

it1 = MyIterator(10)
try:
    while True:
        print(next(it1)) # prints all numbers, but throws error
except StopIteration:
    print("Stop Execution")
it1 = MyIterator(5)
for i in it1:
    print(i)
```
