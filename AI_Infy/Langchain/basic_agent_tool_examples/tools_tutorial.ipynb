```python
from duckduckgo_search import DDGS
from langchain.tools import Tool, DuckDuckGoSearchRun, DuckDuckGoSearchResults
from langchain.agents import initialize_agent, AgentExecutor
from langchain.agents.agent_types import AgentType
from langchain.chat_models import ChatOpenAI, init_chat_model
import torch
from transformers import pipeline
from langchain_community.llms import HuggingFacePipeline
from langchain.agents import Tool
from langchain.tools import DuckDuckGoSearchRun
from langchain_experimental.utilities import PythonREPL
from langchain.utilities import WikipediaAPIWrapper

from AI_Infy.Langchain.pdf_qanda.GeminiLoader import llm
```

```python

search = DuckDuckGoSearchRun()
python_repl = PythonREPL()
wikipedia = WikipediaAPIWrapper()

tools = [
     Tool(
        name='DuckDuckGo Search',
        func= search.run,
        description="Useful for when you need to do a search on the internet to find information that another tool can't find. be specific with your input."
    ),
    Tool(
        name = "python repl",
        func=python_repl.run,
        description="useful for when you need to use python to answer a question. You should input python code"
    ),
    Tool(
        name='wikipedia',
        func= wikipedia.run,
        description="Useful for when you need to look up a topic, country or person on wikipedia"
    )
]
```

```python
from langchain.agents import initialize_agent
tool_names = [tool.name for tool in tools]
prompt = """

You have access to the following tools:
{tools}

Question: {input}
Thought:{agent_scratchpad}
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
(this Thought/Action/Action Input/Observation can repeat N times)
Final Answer: the final answer to the original input question
Begin!

"""

zero_shot_agent = initialize_agent(
    agent="zero-shot-react-description",
    tools=tools,
    agent_kwargs={
        "prefix": prompt,
    },
    stop=['\nFinal Answer'],
    llm=llm,
    verbose=True,
    handle_parsing_errors=True
)
```

```python
zero_shot_agent.run({"input": "When was Barak Obama born and how old is he as of now also show how many years he ruled as president of USA","tools": tools, "tool_names": [tool.name for tool in tools]})
```
