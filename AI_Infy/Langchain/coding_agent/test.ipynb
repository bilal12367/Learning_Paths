```python
import pymupdf4llm
import pathlib

md_text = pymupdf4llm.to_markdown('express_book.pdf')
# pathlib.Path('test.md').write_bytes(md_text.encode())
```

```python
import sys
sys.path.append('../')

from GeminiLoader import llm
```

```python
from langchain.text_splitter import MarkdownHeaderTextSplitter

headers_to_split = [
    ('##','Header 2')
]

splitter  = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split)

docs = splitter.split_text(md_text)
docs
```

```python
filtered_docs = [(doc.metadata['Header 2'], doc) for doc in docs if 'Header 2' in doc.metadata]
ids, txt = zip(*filtered_docs) if filtered_docs else ([], [])
ids, txt
```

```python
from langchain.embeddings import SentenceTransformerEmbeddings

embeddings = SentenceTransformerEmbeddings(model_name='all-MiniLM-L6-v2')

embds = embeddings.embed_documents(texts=list(txt))

embds
```

```python
from langchain.vectorstores import Chroma

db = Chroma(collection_name='express', embedding_function=embeddings, persist_directory='./vc')

db.add_documents(documents=txt, ids=ids)
```

```python
results = db.similarity_search(
    "Routing",
    k=2)

results
```
