```python
from langchain_community.document_loaders import PyMuPDFLoader

loader = PyMuPDFLoader('express_book.pdf')

docs = loader.load()
docs
```

```python
from langchain.text_splitter import SentenceTransformersTokenTextSplitter

list_of_topics = []
for doc in docs:
    if 'Table of Contents' in doc.page_content:
        list_of_topics = doc.page_content.split('\n')[1:-1]
        docs.remove(doc)
```

```python
for doc in docs:
    for topic in list_of_topics:
        if topic in doc.page_content:
            print('========================')
            print(f"Topic: {topic}")
            print(f"Content: {doc.page_content}")
```
