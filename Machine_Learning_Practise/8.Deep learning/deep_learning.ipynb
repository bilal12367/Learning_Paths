```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import  OneHotEncoder, LabelEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split

data = pd.read_csv('Churn_Modelling.csv')
custom_data = np.array([[600, 'France', 'Male', 40, 3, 60000, 2, 1, 1, 5000]],dtype=object)

X = data.iloc[:,3:-1].values
Y = data.iloc[:,-1].values

sc = StandardScaler()
lblEncoder = LabelEncoder()
hotEncoder = ColumnTransformer(transformers=[('encoder',OneHotEncoder(), [1])],remainder='passthrough')
def encoder(X, hotEncoder,lblEncoder, sc, fit):
    if fit == True:
        X[:, 2] = lblEncoder.fit_transform(X[:, 2])
        X = np.array(hotEncoder.fit_transform(X))
        X = sc.fit_transform(X)
    else: 
        X[:, 2] = lblEncoder.transform(X[:, 2])
        X = np.array(hotEncoder.transform(X))
        X = sc.transform(X)
    return X
X = encoder(X,hotEncoder,lblEncoder,sc, True)

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.4, random_state=1)



```

```python
ann = tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=6, activation='relu'),
    tf.keras.layers.Dense(units=6, activation='relu'),
    tf.keras.layers.Dropout(rate=0.2),
    tf.keras.layers.Dense(units=1, activation='sigmoid')
])

ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
ann.fit(x_train, y_train, batch_size=32, epochs=100)
```

```python
custom_data1 = encoder(custom_data, hotEncoder, lblEncoder, sc, False)

print(ann.predict(custom_data1) > 0.5)
```
