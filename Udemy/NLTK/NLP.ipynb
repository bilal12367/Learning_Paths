```python
corpus = """Hello Welcome, to NLP tutorials.
Please do watch the entire course to become expert in NLP.
"""
```

```python
## Toekinization

from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
from nltk.tokenize import word_tokenize,wordpunct_tokenize,TreebankWordTokenizer
```

```python
docs = []
for sentence in documents:
    docs.append(word_tokenize(sentence))
docs
```

#### Stemming

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

##### Porter Stemmer

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

for word in words:
    print(word+'----------->>'+stemmer.stem(word))
```

```python
stemmer.stem("Congratulations")
```

```python
stemmer.stem("Sitting")
```

#### Regexp Stemmer

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|es$|able$', min=4)
```

```python
reg_stemmer.stem("Fuss")
```

#### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
snow_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+' ==========> '+snow_stemmer.stem(word))
```

```python
stemmer.stem('fairly'), stemmer.stem('sportingly')
```

```python
snow_stemmer.stem('fairly'), snow_stemmer.stem('sportingly')
```
